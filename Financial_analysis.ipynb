```python
import pandas as pd

# 1. Create the dataset directly
data = {
    'Company': [
        'Microsoft', 'Microsoft', 'Microsoft',
        'Tesla', 'Tesla', 'Tesla',
        'Apple', 'Apple', 'Apple'
    ],
    'Fiscal Year': [
        2024, 2023, 2022,
        2023, 2022, 2021,
        2024, 2023, 2022
    ],
    'Total Revenue': [
        245122, 211915, 198270,
        96773, 81462, 53823,
        391035, 383285, 394328
    ],
    'Net Income': [
        88136, 72361, 72738,
        14997, 12556, 5519,
        93736, 96995, 99803
    ],
    'Total Assets': [
        512163, 411976, 364840,
        106618, 82338, 62131,
        364980, 352583, 352755
    ],
    'Total Liabilities': [
        243686, 205753, 198298,
        43009, 37634, 31942,
        308030, 290437, 302083
    ],
    'Cash Flow Ops': [
        118260, 87582, 89035,
        13256, 14724, 11497,
        118254, 110543, 122151
    ]
}

# 2. Load data into DataFrame
df = pd.DataFrame(data)

# Sort by Company and Year to ensure correct calculation
df = df.sort_values(by=['Company', 'Fiscal Year'])

# 3. Analyze trends: Calculate Year-Over-Year (YoY) Growth
# We use pct_change() which calculates the percentage change between the current and a prior element.
df['Revenue Growth (%)'] = df.groupby(['Company'])['Total Revenue'].pct_change() * 100
df['Net Income Growth (%)'] = df.groupby(['Company'])['Net Income'].pct_change() * 100
df['Assets Growth (%)'] = df.groupby(['Company'])['Total Assets'].pct_change() * 100
df['Liabilities Growth (%)'] = df.groupby(['Company'])['Total Liabilities'].pct_change() * 100

# Fill NaN values (the first year for each company) with 0 or - for cleaner viewing
df.fillna(0, inplace=True)

# Format the percentage columns for better readability
pd.options.display.float_format = '{:,.2f}'.format

# 4. Display the analyzed data
print("Financial Analysis Summary (Amounts in Millions USD):")
display(df)

# 5. Summarize Findings (Programmatic Summary)
print("\n--- Key Insights ---")
for company in df['Company'].unique():
    company_data = df[df['Company'] == company]
    latest_year = company_data.iloc[-1]
    
    print(f"\n{company} ({latest_year['Fiscal Year']}):")
    print(f" - Revenue Growth: {latest_year['Revenue Growth (%)']:.2f}%")
    print(f" - Net Income Growth: {latest_year['Net Income Growth (%)']:.2f}%")
    
    if latest_year['Net Income Growth (%)'] > 0:
        print(" -> Positive profitability trend observed.")
    else:
        print(" -> Profitability declined compared to previous year.")
```
