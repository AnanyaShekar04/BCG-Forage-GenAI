```python
import pandas as pd

# 1. Define the Data (The knowledge base for the chatbot)
# We use the same data you analyzed in Task 1
financial_data = {
    'Microsoft': {
        '2024': {'Revenue': '$245,122 million', 'Net Income': '$88,136 million', 'Growth': 'positive'},
        '2023': {'Revenue': '$211,915 million', 'Net Income': '$72,361 million', 'Growth': 'stable'}
    },
    'Tesla': {
        '2023': {'Revenue': '$96,773 million', 'Net Income': '$14,997 million', 'Growth': 'stabilizing'},
        '2022': {'Revenue': '$81,462 million', 'Net Income': '$12,556 million', 'Growth': 'explosive'}
    },
    'Apple': {
        '2024': {'Revenue': '$391,035 million', 'Net Income': '$93,736 million', 'Growth': 'mature'},
        '2023': {'Revenue': '$383,285 million', 'Net Income': '$96,995 million', 'Growth': 'stable'}
    }
}

# 2. Define the Chatbot Function (Rule-Based Logic)
def simple_chatbot(user_query):
    # Normalize the query to lower case to make matching easier
    query = user_query.lower()

    # LOGIC 1: Check if the user is asking about a specific company
    company = None
    if 'microsoft' in query:
        company = 'Microsoft'
    elif 'tesla' in query:
        company = 'Tesla'
    elif 'apple' in query:
        company = 'Apple'
    else:
        return "Please specify which company you are interested in (Microsoft, Tesla, or Apple)."

    # LOGIC 2: Check what metric the user wants (Revenue vs Net Income)
    latest_year = list(financial_data[company].keys())[0] # Gets the most recent year available
    data = financial_data[company][latest_year]

    if 'total revenue' in query or 'revenue' in query:
        return f"The Total Revenue for {company} for fiscal year {latest_year} was {data['Revenue']}."
    
    elif 'net income' in query or 'profit' in query:
        return f"The Net Income for {company} for fiscal year {latest_year} was {data['Net Income']}."
    
    elif 'growth' in query or 'trend' in query:
         return f"{company} has shown {data['Growth']} growth trends in recent years."
        
    else:
        return "Sorry, I can only provide information on 'Total Revenue', 'Net Income', and 'Growth'."

# 3. Test the Chatbot (Automated Test)
# This simulates a user asking questions so you don't have to type manually.

print("--- Automated Chatbot Test ---")

# Test 1: Ask about Apple
print("User: What is the total revenue for Apple?")
print("Chatbot:", simple_chatbot("What is the total revenue for Apple?"))
print("-" * 30)

# Test 2: Ask about Microsoft
print("User: Tell me the net income for Microsoft")
print("Chatbot:", simple_chatbot("Tell me the net income for Microsoft"))
print("-" * 30)

# Test 3: Ask about Tesla
print("User: How is Tesla growth?")
print("Chatbot:", simple_chatbot("How is Tesla growth?"))
print("-" * 30)

# Test 4: Ask about an unknown company (to test error message)
print("User: Revenue for Google")
print("Chatbot:", simple_chatbot("Revenue for Google"))
```
